```python
# 1. Write a function to normalize a list of numbers between 0 and 1
nums = [3, 6, 8, 11, 32, 6, 83]

def normalise(nums: list):
    # Handle empty lists to avoid errors
    if not nums:
        return []
        
    # Find the range (min and max)
    min_val = min(nums)
    max_val = max(nums)
    
    # Avoid division by zero if all numbers in the list are the same
    if max_val == min_val:
        return [0.0] * len(nums)
    
    # Calculate normalized values
    # Python performs float division by default with the / operator
    normalized = [(x - min_val) / (max_val - min_val) for x in nums]
    
    return normalized

print(normalise(nums))
```

```python
# 2. Implement manual matrix multiplication (no NumPy)
def matrix_multiply(A, B):
    # Get dimensions
    rows_A = len(A)
    cols_A = len(A[0])
    rows_B = len(B)
    cols_B = len(B[0])

    # 1. Validation: Check if multiplication is possible
    if cols_A != rows_B:
        raise ValueError("Matrix dimensions incompatible for multiplication.")

    # 2. Initialize the result matrix with zeros
    # Result will be (rows_A x cols_B)
    result = [[0 for _ in range(cols_B)] for _ in range(rows_A)]

    # 3. Perform the multiplication
    for i in range(rows_A):          # Iterate through rows of A
        for j in range(cols_B):      # Iterate through columns of B
            for k in range(cols_A):  # Iterate through columns of A / rows of B
                result[i][j] += A[i][k] * B[k][j]

    return result

# Example Usage:
matrix_a = [
    [1, 2],
    [3, 4]
]

matrix_b = [
    [5, 6],
    [7, 8]
]

print(matrix_multiply(matrix_a, matrix_b))
# Output: [[19, 22], [43, 50]]
```
