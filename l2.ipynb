```python
# 11. Load an image and display it
# 12. Resize image to 224×224
import cv2

# 1. Load the image
# 'path_to_image.jpg' is the file name or full path
img = cv2.imread('cat.jpg')

# print(img)

img = cv2.resize(img, (400, 400))
# 2. Display the image in a window
# The first argument is the window name, the second is the image object
cv2.imshow('Loaded Image', img)

# 3. Wait for a key press to close the window
# 0 means it waits indefinitely until you press any key
cv2.waitKey(0)

# 4. Cleanup: Close all open windows
cv2.destroyAllWindows()

##---------------- .waitKey() , .destroyAllWindows(), .resize(), .imread(), .imshow()
```

```python
# 13. Convert image to grayscale

# The function cv2.cvtColor is used for color space conversions
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Display the result
cv2.imshow('Grayscale', gray_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

##----------------------- .cvtColor()  .COLOR_BGR2GRAY
```

```python
# 14. Normalize pixel values to [0,1]

# 1. Convert to float32 first (to maintain precision during division)
# 2. Divide by 255.0
normalized_img = img.astype('float32') / 255.0
# print(normalized_img)
# Check the values
print("Original Max:", img.max())         # Should be around 255
print("Normalized Max:", normalized_img.max()) # Should be 1.0
```

```python
# 15. Apply Gaussian blur

# The second argument (5, 5) is the 'Kernel Size'
# The third argument is 'sigmaX' (0 means it's calculated automatically based on kernel size)
blurred_img = cv2.GaussianBlur(img, (5, 5), 0)

cv2.imshow('Original', img)
cv2.imshow('Gaussian Blur', blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

##---------- .GaussianBlur()
```

```python
# 16. Apply edge detection (Canny)
import cv2

# 1. Convert to gray (Canny needs a single channel)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Apply Canny Edge Detection
# The two numbers (100, 200) are the Low Threshold and High Threshold
edges = cv2.Canny(gray, 100, 200)

cv2.imshow('Canny Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()

##------- .cvtColor(), .COLORBGR2GRAY, .Canny()

```

```python
# 17. Save processed image

#This will save the 'edges' image we just created
# It automatically detects the format based on the file extension (.jpg, .png, etc.)
cv2.imwrite('my_detected_edges.png', edges)

print("Image saved successfully!")

##------ .imwrite()
```

```python
# 18. Convert between PIL ↔ NumPy ↔ OpenCV formats
import cv2
import numpy as np
from PIL import Image

# --- 1. OpenCV (NumPy) to PIL ---
# Load with OpenCV (BGR format)
cv2_img = img

# Convert BGR to RGB (PIL needs RGB)
img_rgb = cv2.cvtColor(cv2_img, cv2.COLOR_BGR2RGB)

# Create PIL image
pil_img = Image.fromarray(img_rgb)
# pil_img.show()  # This opens the default OS image viewer


# --- 2. PIL to NumPy (OpenCV compatible) ---
# Convert PIL back to NumPy array
new_ndarray = np.array(pil_img)

# Convert RGB back to BGR so OpenCV can show it correctly
final_cv2_img = cv2.cvtColor(new_ndarray, cv2.COLOR_RGB2BGR)

cv2.imshow('Back to OpenCV', final_cv2_img)
cv2.waitKey(0)
cv2.destroyAllWindows()

##-------------- .fromarray(), np.array()
```
