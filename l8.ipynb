```python
import torch
import torchvision.transforms as T
import torchvision.transforms.functional as F
import random
from PIL import Image
import numpy as np

def apply_augmentations(image_tensor):
    """
    Applies consistent augmentations to the 6-channel tensor.
    image_tensor shape: (6, 224, 224)
    """
    
    # 1. Random Horizontal & Vertical Flip
    if random.random() > 0.5:
        image_tensor = F.hflip(image_tensor)
    if random.random() > 0.5:
        image_tensor = F.vfl ip(image_tensor)
        
    # 2. Random Resize and Crop (Simulates different scales)
    # We resize slightly larger and crop back to 224
    if random.random() > 0.5:
        resize = T.RandomResizedCrop(size=(224, 224), scale=(0.8, 1.0))
        image_tensor = resize(image_tensor)

    # 3. Gaussian Blur (Simulates low-quality captures)
    if random.random() > 0.3:
        blurrer = T.GaussianBlur(kernel_size=(5, 9), sigma=(0.1, 2.0))
        image_tensor = blurrer(image_tensor)

    # 4. Compression Simulation (Adding slight Gaussian Noise)
    # JPEG compression is hard to do on a tensor, so we simulate 
    # artifacting by adding a tiny bit of random noise.
    if random.random() > 0.5:
        noise = torch.randn_like(image_tensor) * 0.02
        image_tensor = image_tensor + noise
        image_tensor = torch.clamp(image_tensor, 0, 1)

    return image_tensor
```
